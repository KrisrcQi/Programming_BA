```python
# Install packages
import numpy as np
import pandas as pd
```

```python
# Creating a numpy array

weekly_sales = np.array([100, 200, 300, 400, 500])
print(weekly_sales)
```

```python
weekly_sales
```

```python
# Compute summary of statisitics

mean_value = np.mean(weekly_sales)
print("Mean value:", mean_value)

max_value = np.max(weekly_sales)
print("Max value:", max_value)

min_value = np.min(weekly_sales)
print("Min value:", min_value)

std_value = np.std(weekly_sales)
print("Standard Deviation:", std_value)
print("================================")
print("Summary of statistics:")
print("Mean value:", round(mean_value,2))
print("Max value:", round(max_value,2))
print("Min value:", round(min_value,2))
print("Standard Deviation:", round(std_value,2))
```

```python
# Or, using the simple math function from python
mean_value_0 = weekly_sales.mean()
max_value_0 = weekly_sales.max()
min_value_0 = weekly_sales.min()
std_value_0 = weekly_sales.std()    

print(mean_value_0, max_value_0, min_value_0, std_value_0)
```

```python
# Compute store_level totals

# functions of Aggregation by "groupby" and "agg"
store_level_totals = df_1.groupby("Store", as_index=False)
store_level_totals = store_level_totals.agg(Total_Sales = ("Sales", "sum"), Total_Profit = ("Profit", "sum"))
print(store_level_totals)

store_level_totals_1 = df_1.groupby("Store")
store_level_totals_1 = store_level_totals_1.agg(Total_Sales = ("Sales", "sum"), Total_Profit = ("Profit", "sum"))

print("================================")
print(store_level_totals_1)

# as_index = False means that the group keys will not be used as the index of the resulting DataFrame. 
# Instead, the group keys will be included as regular columns in the output DataFrame. 
# This is useful when you want to keep the original structure of the DataFrame and avoid having the group keys as the index.

```

```python
# Sorting the values by certain order and columns
store_level_totals.sort_values(by="Total_Sales", ascending=False) # default is ascending
print(store_level_totals)
```

```python
# Combining all codes together
store = (df_1.groupby("Store", as_index=False)
         .agg(Total_Sales = ("Sales", "sum"), 
              Total_Profit = ("Profit", "sum"))
         .sort_values(by="Total_Sales", ascending=False))
store
```
