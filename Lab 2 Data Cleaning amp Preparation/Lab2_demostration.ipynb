```python
# checking for missing values

print(df.isnull())
print("=" * 50, "\n showing the number of missing values in each column:")
print(df.isnull().sum())
```

```python
# show rows where Region is missing:

df[df["Region"].isnull()] # or using .isna()
# df[df["Region"].isna()]
```

```python
# Filling all missing values in the Region column with "Unknown"
df["Region"] = df["Region"].fillna("Central")

# verify the missing values have been filled

df["Region"].isnull() # see different from with df[]
# df[df["Region"].isnull()]

"""
Or see the number of missing values in the Region column:
df["Region"].isnull().sum()
"""
```

```python
# checking for the duplication rows in the dataset:

df.duplicated()
```

```python
# sum of the duplication rows in the dataset:
print("The number of duplicated rows in the dataset:")
print(df.duplicated().sum())
print("=" * 50)

print("\n","Show the duplicated rows in the dataset:")
print(df[df.duplicated()])
```

```python
# remove the duplicated rows in the dataset:
df = df.drop_duplicates()

# verify the duplicated rows have been removed:
print("\n","The number of duplicated rows in the dataset after removing duplicates:")
print(df.duplicated().sum())

# check the data shape after removing duplicates:
print("\n","Data shape after removing duplicates:")
print(df.shape)
```

```python
# convert the date column to datetime format:
## Method 1:
"""prase_function

pd.read_csv(path, parse_dates=["date"])

"""

## Method 2:
"""datetime_function

pd.to_datetime(df["date"], format="%Y-%m-%d")

"""

# df.info()
df["Date"] = pd.to_datetime(df["Date"], format="%d/%m/%Y")
df.info()
```

```python
# create a new column for the year and month of the transaction:
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month

# create a name of the month column:
df["Month_name"] = df["Date"].dt.month_name()
df["Month_name_1"] = df["Date"].dt.strftime("%b")
```

```python
df.head()
```

```python
# grouping and aggregating practices:
"""
1. which region makes the most revenue (sale)?
2. which category is most profitable?
"""

# sales_region_1 of most revenue:
sales_region_1 = (
    df.groupby("Region")[["Sales", "Profit"]]
    .sum()
    .sort_values(by="Sales", ascending=False)
)
print(sales_region_1, "\n", "=" * 50)

# sales_region_2 of most revenue:
sales_region_2 = (
    df.groupby("Region", as_index=False)
    .agg({"Sales": "sum", "Profit": "sum"})
    .sort_values(by="Sales", ascending=False)
)
print(sales_region_2, "\n", "=" * 50)

# sales_category_3 of most revenue:
sales_category_3 = (
    df.groupby("Region", as_index=False)
    .agg(Total_Sales = ("Sales", "sum"), Total_Profit = ("Profit", "sum"))
    .sort_values(by="Total_Sales", ascending=False)
)
print(sales_category_3)
```

```python
# profit_cat of most profitable:

profit_cat = (
    df.groupby("Category", as_index=False)
    .agg(Total_Sales = ("Sales", "sum"), Total_Profit = ("Profit", "sum"))
    .sort_values(by="Total_Profit", ascending=False)
)
print(profit_cat, "\n", "=" * 50)
print("Top 3 most profitable categories:")
print(profit_cat.head(3))
```
