### Dataset: product_performance.csv

```python
import numpy as np
import pandas as pd
```

```python
# import and inspect dataset

df = pd.read_csv("product_performance.csv")

print("First_5:", "\n", df.head(5), "\n", "=" * 50, "\n",
      "Data_shape:", "\n", df.shape, "\n", "=" * 50, "\n",
      "Data_statistics:", "\n", df.describe(), "\n", "=" * 50, "\n")

print(df.info())
```

```python
# create a Profit_margin column

df["Profit_margin"] = df["Profit"]/df["Sales"]

# create a new table with all numerical columns 
df_num = df[["Sales", "Cost", "Profit", "Profit_margin"]]
display(df_num.head(5))
```

```python
# flag high vs low profit products:

"""steps
1. looking for where profit is greater or lower than benchmark 
    (0.25 as default, it can be changed later)
2. create a new table with product name and profit_margin, 
    be ready for make flag as High and Low
3. count the number of both levels of profit margin
"""

df["Profit_flag"] = np.where(df["Profit_margin"] >= 0.25, "High", "Low")

df_products = df[["Product", "Profit_margin", "Profit_flag"]]
display(df_products)

print(df["Profit_flag"].value_counts())
```

```python
# flag median sales: 

"""steps
1. calculate median of sales, assigning to "median_sales"
2. looking for where sales greater or lower than median_sales, saving into the new columns as "Sales_level"
3. create a new table with product name, sales, sales_level
4. count the number of both sales levels
"""

median_sales = df["Sales"].median()

df["Sales_level"] = np.where(df["Sales"] >= median_sales, "Above_med", "Below_med")
df_sales_level = df[["Product", "Sales", "Sales_level"]]
display(df_sales_level)

print(df_sales_level["Sales_level"].value_counts())
# therefore, it should be a half-half level of sales, but students can change this to the mean, which will return the a distribution
```

```python
# looping over categories:

uni_cat = df["Category"].unique()

# loop each category from this unique_cat list: will return a 1x4 vector
# for i in uni_cat:
#     print(i)

# looking for how many the products within each category:
## in another word, sum products by category:

"""groupby method
num_product_cat = (
    df.groupby("Category")["Product"].sum()
)
print(num_product_cat)

However, product is str, not float
"""

"""loop mechanism interpration
subset = df[df["Category"] == i]
change i to either of category value in the dataset

"""

for i in uni_cat:
    subset = df[df["Category"] == i]
    count_products = len(subset)
    print("Category:", i, "| number of products:", count_products)
```

```python
# average profit per category:
"""groupby
avg_profit = (
    df.groupby("Category").agg(avg_profit =("Profit", "mean"))
)
avg_profit
"""

for cat in uni_cat:
    subset = df[df['Category'] == cat]
    avg_profit = subset['Profit'].mean()
    print('Category:', cat, '| Avg Profit:', avg_profit)
```

```python
# functions and applying functions

## define a functions

def _print (x):
    return print(x)

def margin_check (margin):
    if margin >= 0.25:
        return "High"
    else:
        return "Low"

# margin_check(1)
# margin_check(df_products["Profit_margin"])
df_products["Profit_margin"].apply(margin_check)
```

```python
# create a promotion flag for profit >= 80

df["Promotion"] = np.where(df["Profit"] >= 80, "Consider_Promotion", "No_Need")

promotion_sum = (
    df.groupby(["Category", "Promotion"])["Product"].count()
)
display(promotion_sum)
```

```python
# three level of conditions:
df["Profit_category"] = np.where(
    df["Profit_margin"] >= 0.30, "Very_high",
    np.where(
        df["Profit_margin"] >= 0.15, "Medium",
        "Low"
    )
)
df_profit_cat = df[["Category", "Profit", "Profit_category"]]
print(df_profit_cat)

## method 2 in functions
def profit_cate (profit):
    if profit >= 0.30:
        return "Very_high"
    elif profit >= 0.15:
        return "Medium"
    else:
        return "Low"

df["Profit_margin"].apply(profit_cate)
```

```python
# change the input for function as whole dataset:
"""logic
#   High_Risk if Profit_Margin < 0.15 AND Sales > median_sales
#   Low_Risk otherwise
"""

def risk_flag (data):
    if (data["Profit_margin"] < 0.15) and (data["Sales"] > median_sales):
        return "High_Risk"
    else:
        return "Low_Risk"

df["Risk_Flag"] = df.apply(risk_flag, axis=1)
df["Risk_Flag"].value_counts()

```
